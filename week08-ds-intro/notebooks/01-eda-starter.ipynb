## 4. Aykırı değerler — IQR yöntemi

```python
def iqr_outliers(series):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return series[(series < lower) | (series > upper)]

iqr_outliers(df['age'].dropna())
```
